```python
import pandas as pd

df = pd.read_csv('depths141 (2).csv')
df2 = pd.read_csv('depths188 (2).csv')
df3 = pd.read_csv('depths197 (2).csv')
df4 = pd.read_csv('depths211 (2).csv')

finaldf = pd.concat([df[['origin_time']], df2[['origin_time']], df3[['origin_time']], df4[['origin_time']]]).reset_index()[['origin_time']]
finaldf['event_cnt'] = 1
finaldf['event_cnt'] = finaldf.event_cnt.cumsum()
finaldf.to_csv('ba1a_fractures.csv', index=False)
```
